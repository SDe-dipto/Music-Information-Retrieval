```python
onset_frames = librosa.onset.onset_detect(x, sr=sr)
print(onset_frames)
onset_times = librosa.frames_to_time(onset_frames, sr=sr)
print(onset_times)
onset_samples = librosa.frames_to_samples(onset_frames)
print(onset_samples)
```
