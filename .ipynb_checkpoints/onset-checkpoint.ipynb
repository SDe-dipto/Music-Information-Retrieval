```python
%matplotlib inline
import numpy, scipy, matplotlib.pyplot as plt, IPython.display as ipd
import librosa, librosa.display
```

```python
plt.style.use('seaborn-muted')
plt.rcParams['figure.figsize'] = (14, 5)
plt.rcParams['axes.grid'] = True
plt.rcParams['axes.spines.left'] = False
plt.rcParams['axes.spines.right'] = False
plt.rcParams['axes.spines.bottom'] = False
plt.rcParams['axes.spines.top'] = False
plt.rcParams['axes.xmargin'] = 0
plt.rcParams['axes.ymargin'] = 0
plt.rcParams['image.cmap'] = 'gray'
plt.rcParams['image.interpolation'] = None
```

```python
x,sr=librosa.load('T39-piano.wav')
```

```python
ipd.Audio(x,rate=sr)
```

```python
plt.figure(figsize=(14, 5))
librosa.display.waveshow(x, sr)
```

```python
plt.figure(figsize=(14, 5))
librosa.display.waveplot(x, sr=22050)
```

```python
plt.figure(figsize=(14, 5))
librosa.display.waveshow(x, sr=22050)
```

```python
onset_frames = librosa.onset.onset_detect(x, sr=sr, wait=1, pre_avg=1, post_avg=1, pre_max=1, post_max=1)
print(onset_frames)
```

```python
onset_times = librosa.frames_to_time(onset_frames)
print(onset_times)
```

```python
S = librosa.stft(x)
logS = librosa.amplitude_to_db(abs(S))
```

```python
plt.figure(figsize=(14, 5))
librosa.display.specshow(logS, sr=sr, x_axis='time', y_axis='log', cmap='Reds')
plt.vlines(onset_times, 0, 10000, color='#3333FF')
```

```python
plt.figure(figsize=(14, 5))
librosa.display.waveshow(x, sr=sr)
plt.vlines(onset_times, -0.8, 0.79, color='g', alpha=0.8) 
```

```python
clicks = librosa.clicks(frames=onset_frames, sr=sr, length=len(x))
```

```python
ipd.Audio(x + clicks, rate=sr)
```

```python
ipd.Audio(numpy.vstack([x, clicks]), rate=sr)
```

```python
x1,sr1=librosa.load('T08-violin.wav')
ipd.Audio(x1,rate=sr1)
```

```python
plt.figure(figsize=(14, 5))
librosa.display.waveshow(x1, sr=sr1)
plt.vlines(onset_times, -0.8, 0.79, color='r', alpha=0.8) 
```
