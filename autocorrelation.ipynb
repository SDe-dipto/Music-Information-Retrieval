```python
r = numpy.correlate(x, x, mode='full')[len(x)-1:]
print(x.shape, r.shape)
```

```python
plt.figure(figsize=(14, 5))
plt.plot(r[:10000])
plt.xlabel('Lag (samples)')
plt.xlim(0, 10000)
```

```python
plt.figure(figsize=(14, 5))
plt.plot(r[:20000])
plt.xlabel('Lag (samples)')
plt.xlim(0, 20000)
```

```python
plt.figure(figsize=(14, 5))
plt.plot(r[:5000])
plt.xlabel('Lag (samples)')
plt.xlim(0, 5000)
```

```python
r = librosa.autocorrelate(x, max_size=5000)
plt.figure(figsize=(14, 5))
plt.plot(r[:200])
```

```python
r = librosa.autocorrelate(x, max_size=5000)
plt.figure(figsize=(14, 5))
plt.plot(r[:1000])
```

```python
midi_hi = 120.0
midi_lo = 12.0
f_hi = librosa.midi_to_hz(midi_hi)
f_lo = librosa.midi_to_hz(midi_lo)
t_lo = sr/f_hi
t_hi = sr/f_lo
```

```python
print(f_lo, f_hi)
print(t_lo, t_hi)
```

```python
r[:int(t_lo)] = 0
r[int(t_hi):] = 0
```

```python
plt.figure(figsize=(14, 5))
plt.plot(r[:1400])
```

```python
t_max = r.argmax()
print(t_max)
```
